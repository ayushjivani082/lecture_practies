### what is pandas?
###### A Python library to work with table data (excel) using code.


```python
import numpy as py
import pandas as pd
```

```python
marks = pd.Series([67 , 77 , 89 , 95])
print(marks)
```

```python
marks = pd.Series([56 , 78 , 89 , 82] , index = ["amit" , "bhargav", "man" , "ved"] ,  name="Marks")

print(marks)
```

```python
# by lable

print("man's Marks:" , marks["man"])

# by index

print("First Student:" , marks.iloc[3])
```

```python
fees = pd.Series({"C++":60000 , "Python":90000 , "Java":100000})

print(fees)
```

```python
print(marks + 5)

print()

print(marks[marks >= 80])

print()

print("Max:" , marks.max())

print("Min:" , marks.min())

print("Mean:" , marks.mean())
```

```python
# Dictionary of lists


students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
    'course': ['Computer Science', 'Mathematics', 'Physics', 'Biology', 'Biology'],
    'marks': [85, 92, 78, 88, 95],
    'fees': [5000.00, 4500.00, 5000.00, 6000.00, 5500.00]
})

print(students)
```

```python
data = [
    [101, "man", 25000],
    [102, "ved", 30000],
    [103, "prit", 28000]
]

columns = ["ID", "Name", "Salary"]

employee = pd.DataFrame(data , columns=columns)

print(employee)
```
