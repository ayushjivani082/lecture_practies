```python
import numpy as np

print(np.__version__)
```

```python
python_list = [1 , 2 , 3 , 4 , 5]

doubled_list = [x * 2 for x in python_list]

print(doubled_list)

print("\n=====  Numpy Array ======\n")

numpy_array = np.array([1 , 2 , 3 , 4 , 5])

doubled_array = numpy_array * 2

print(doubled_array)
```

```python
# Create a 1D array of integer from 0 to 9

arr = np.arange(0 , 10 , 2)

print(arr)
print(arr.shape)
```

```python
# Generate a 2D Array with random values and print shape and datatypes

arr = np.random.rand(3 , 4)

print(arr)

print(type(arr))

print(arr.shape)
```

```python
# Create 50 Evenly spaced values between 0 to 10

arr = np.linspace(0 , 10 , 2)

print(arr)
```

```python
# np.ones

arr = np.ones((3 , 4))

print(arr)
```

```python
arr = np.random.randint(1 , 21 , size=(4 , 4))

print(arr)
```

```python
arr = np.arange(1 , 16)

print(arr)

print("Sliced Array:" , arr[3:9])
print("Sliced Array:" , arr[::-1])
print("Sliced Array:" , arr[-3:])
print("Sliced Array:" , arr[:5])
```

```python

arr = np.arange(1 , 10).reshape(3 , 3)

print(arr)

print("Element:" , arr[1 , 2])
```

```python
arr = np.arange(1 , 17).reshape(4 , 4)

print(arr)

print("Slice Matrix:" , arr[-2: , -2:])
```
